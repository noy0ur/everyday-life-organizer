```python
# EVERYDAY LIFE ORGANIZER
# Step 1: Import libraries

import pandas as pd
from datetime import date

print("Welcome to Everyday Life Organizer!")
print("Your daily productivity dashboard is ready.")
print("Today's date:", date.today())
```

```python
# STEP 2: DAILY TASK MANAGER

import pandas as pd
from datetime import date

# Store tasks in a list
tasks = []

def add_task(title, priority="Medium"):
    task = {
        "Task": title,
        "Priority": priority,
        "Status": "Pending",
        "Date Added": str(date.today())
    }
    tasks.append(task)
    print(f"Task added successfully: {title}")

def view_tasks():
    if not tasks:
        print("No tasks available.")
    else:
        display(pd.DataFrame(tasks))

def complete_task(task_number):
    if 1 <= task_number <= len(tasks):
        tasks[task_number - 1]["Status"] = "Completed"
        print("Task marked as completed!")
    else:
        print("Invalid task number.")

# Add sample tasks
add_task("Complete Python project", "High")
add_task("Study machine learning", "Medium")
add_task("Buy groceries", "Low")

print("\nMY TASK LIST")
view_tasks()

# Complete the first task
complete_task(1)

print("\nUPDATED TASK LIST")
view_tasks()
```

```python
# STEP 3: SHOPPING LIST MANAGER

import pandas as pd

shopping_list = []

def add_item(item, quantity=1):
    shopping_list.append({
        "Item": item,
        "Quantity": quantity,
        "Purchased": "No"
    })
    print(f"Added to shopping list: {item}")

def view_shopping_list():
    if shopping_list:
        display(pd.DataFrame(shopping_list))
    else:
        print("Your shopping list is empty.")

def mark_purchased(item_name):
    for item in shopping_list:
        if item["Item"].lower() == item_name.lower():
            item["Purchased"] = "Yes"
            print(f"{item_name} marked as purchased!")
            return

    print("Item not found.")

# Add sample shopping items
add_item("Milk", 2)
add_item("Rice", 1)
add_item("Eggs", 12)
add_item("Vegetables", 3)

print("\nMY SHOPPING LIST")
view_shopping_list()

# Mark an item as purchased
mark_purchased("Milk")

print("\nUPDATED SHOPPING LIST")
view_shopping_list()
```

```python
# STEP 4: BILL PAYMENT TRACKER

import pandas as pd
from datetime import date

bills = []

def add_bill(name, amount, due_date):
    bill = {
        "Bill Name": name,
        "Amount (₹)": amount,
        "Due Date": due_date,
        "Status": "Unpaid"
    }

    bills.append(bill)
    print(f"Bill added successfully: {name}")

def view_bills():
    if bills:
        display(pd.DataFrame(bills))
    else:
        print("No bills recorded.")

def mark_bill_paid(bill_name):
    for bill in bills:
        if bill["Bill Name"].lower() == bill_name.lower():
            bill["Status"] = "Paid"
            print(f"{bill_name} marked as paid!")
            return

    print("Bill not found.")

def show_unpaid_total():
    total = sum(
        bill["Amount (₹)"]
        for bill in bills
        if bill["Status"] == "Unpaid"
    )

    print(f"Total unpaid bills: ₹{total:,.2f}")

# Add sample bills
add_bill("Electricity", 850, "2026-10-25")
add_bill("Internet", 599, "2026-10-28")
add_bill("Mobile Recharge", 299, "2026-10-30")

print("\nMY BILLS")
view_bills()

print("\nUNPAID TOTAL")
show_unpaid_total()

# Mark one bill as paid
mark_bill_paid("Internet")

print("\nUPDATED BILLS")
view_bills()

show_unpaid_total()
```

```python
# STEP 5: PERSONAL NOTES MANAGER

import pandas as pd
from datetime import datetime

notes = []

def add_note(title, content):
    note = {
        "Title": title,
        "Content": content,
        "Created At": datetime.now().strftime("%Y-%m-%d %H:%M")
    }

    notes.append(note)
    print(f"Note saved successfully: {title}")

def view_notes():
    if notes:
        display(pd.DataFrame(notes))
    else:
        print("No notes available.")

def search_notes(keyword):
    results = [
        note for note in notes
        if keyword.lower() in note["Title"].lower()
        or keyword.lower() in note["Content"].lower()
    ]

    if results:
        display(pd.DataFrame(results))
    else:
        print("No matching notes found.")

# Add sample notes
add_note(
    "Python Learning",
    "Practice Python functions and database operations."
)

add_note(
    "Shopping Reminder",
    "Buy milk, rice, and vegetables."
)

add_note(
    "Career Goal",
    "Prepare a Python developer resume and apply for jobs."
)

print("\nMY PERSONAL NOTES")
view_notes()

print("\nSEARCH RESULTS: PYTHON")
search_notes("Python")
```

```python
# STEP 6: EVERYDAY LIFE ORGANIZER DASHBOARD

import pandas as pd

# Task summary
total_tasks = len(tasks)

completed_tasks = sum(
    1 for task in tasks
    if task["Status"] == "Completed"
)

pending_tasks = total_tasks - completed_tasks

# Shopping summary
total_items = len(shopping_list)

purchased_items = sum(
    1 for item in shopping_list
    if item["Purchased"] == "Yes"
)

remaining_items = total_items - purchased_items

# Bill summary
total_bills = len(bills)

unpaid_amount = sum(
    bill["Amount (₹)"]
    for bill in bills
    if bill["Status"] == "Unpaid"
)

# Notes summary
total_notes = len(notes)

# Display dashboard
dashboard = pd.DataFrame({
    "Metric": [
        "Total Tasks",
        "Completed Tasks",
        "Pending Tasks",
        "Shopping Items",
        "Purchased Items",
        "Remaining Shopping Items",
        "Total Bills",
        "Unpaid Bills Amount (₹)",
        "Saved Notes"
    ],
    "Value": [
        total_tasks,
        completed_tasks,
        pending_tasks,
        total_items,
        purchased_items,
        remaining_items,
        total_bills,
        unpaid_amount,
        total_notes
    ]
})

print("===================================")
print("   EVERYDAY LIFE ORGANIZER")
print("===================================")

display(dashboard)
```

```python
# STEP 7: VISUAL ANALYTICS

import matplotlib.pyplot as plt

# Chart 1: Task progress
completed = sum(
    1 for task in tasks
    if task["Status"] == "Completed"
)

pending = len(tasks) - completed

plt.figure(figsize=(7, 4))
plt.bar(
    ["Completed", "Pending"],
    [completed, pending]
)
plt.title("Daily Task Progress")
plt.ylabel("Number of Tasks")
plt.show()


# Chart 2: Shopping progress
purchased = sum(
    1 for item in shopping_list
    if item["Purchased"] == "Yes"
)

not_purchased = len(shopping_list) - purchased

plt.figure(figsize=(7, 4))
plt.pie(
    [purchased, not_purchased],
    labels=["Purchased", "Not Purchased"],
    autopct="%1.1f%%"
)
plt.title("Shopping Progress")
plt.show()


# Chart 3: Bill payment status
paid = sum(
    1 for bill in bills
    if bill["Status"] == "Paid"
)

unpaid = len(bills) - paid

plt.figure(figsize=(7, 4))
plt.bar(
    ["Paid", "Unpaid"],
    [paid, unpaid]
)
plt.title("Bill Payment Status")
plt.ylabel("Number of Bills")
plt.show()
```
